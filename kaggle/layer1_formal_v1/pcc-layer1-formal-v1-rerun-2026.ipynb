```python
from pathlib import Path
import json, subprocess, torch
assert torch.cuda.is_available(), 'GPU_REQUIRED: no CUDA device'
assert torch.cuda.get_device_capability(0) == (6,0), torch.cuda.get_device_capability(0)
repo=Path('/tmp/PCC')
subprocess.run(['git','clone','--quiet','https://github.com/changxinjiresearch/PCC.git',str(repo)],check=True)
subprocess.run(['git','-C',str(repo),'checkout','--quiet','7f2a7fd'],check=True)
nb=json.loads((repo/'archive/pcc-experiments-original.ipynb').read_text())
def clean_source(index, limit=None):
    lines=''.join(nb['cells'][index]['source']).splitlines()
    if limit is not None: lines=lines[:limit]
    return '\n'.join(line for line in lines if not line.lstrip().startswith('!'))
exec(compile(clean_source(73), 'archive-cell-73', 'exec'), globals())
exec(compile(clean_source(74,437), 'archive-cell-74-definitions', 'exec'), globals())
exec(compile(clean_source(75), 'archive-cell-75-formal-v1', 'exec'), globals())
formal=Path('/kaggle/working/pcc_independent_baseline/stage2_pcc_guided_student_learning/layer1_current_segmentation_FORMAL_v1_5fold')
case_metrics=formal/'layer1_FORMAL_v1_case_metrics.csv'
assert case_metrics.exists()
assert len(__import__('pandas').read_csv(case_metrics)) == 40
assert len(list((formal/'checkpoints').glob('baseline_seg_fold_*_FORMAL_v1.pt'))) == 5
assert len(list((formal/'checkpoints').glob('pcc_seg_corrector_fold_*_FORMAL_v1.pt'))) == 5
(formal/'LAYER1_COMPLETION.json').write_text(json.dumps({'status':'COMPLETE','cases':40,'folds':5,'gpu_required':True,'source_cells':[73,74,75],'cell74_definition_lines':437,'repository_commit':'7f2a7fd'},indent=2)+'\n')
```
