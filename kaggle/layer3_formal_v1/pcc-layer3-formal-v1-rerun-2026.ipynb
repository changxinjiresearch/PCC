```python
from pathlib import Path
import json, subprocess, torch
assert torch.cuda.is_available(), 'GPU_REQUIRED: no CUDA device'
assert torch.cuda.get_device_capability(0) == (6,0), torch.cuda.get_device_capability(0)
repo=Path('/tmp/PCC')
subprocess.run(['git','clone','--quiet','https://github.com/changxinjiresearch/PCC.git',str(repo)],check=True)
subprocess.run(['git','-C',str(repo),'checkout','--quiet','f053d57'],check=True)
nb=json.loads((repo/'archive/pcc-experiments-original.ipynb').read_text())
old='/kaggle/input/datasets/jeechangxin/layer1-formal-v1-5fold-results-backup/'
new='/kaggle/input/pcc-layer1-formal-v1-rerun-2026/'
def source(index):
    text=''.join(nb['cells'][index]['source'])
    assert old in text
    return text.replace(old,new)
exec(compile(source(84), 'archive-cell-84-layer3a-formal', 'exec'), globals())
exec(compile(source(85), 'archive-cell-85-layer3b-formal', 'exec'), globals())
a=Path('/kaggle/working/Layer3_FORMAL_v1_40case_pathology_reliance_occlusion')
b=Path('/kaggle/working/Layer3B_FORMAL_v1_40case_correction_localization')
assert len(__import__('pandas').read_csv(a/'Layer3_FORMAL_v1_case_metrics.csv')) == 40
assert len(__import__('pandas').read_csv(b/'Layer3B_FORMAL_v1_case_metrics.csv')) == 40
(Path('/kaggle/working')/'LAYER3_COMPLETION.json').write_text(json.dumps({'status':'COMPLETE','layer3a_cases':40,'layer3b_cases':40,'gpu_required':True,'source_cells':[84,85],'repository_commit':'f053d57','scientific_code_changes':False,'path_relocation_only':True},indent=2)+'\n')
```
