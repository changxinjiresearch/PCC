```python
from pathlib import Path
import subprocess,sys,json
repo=Path('/tmp/PCC')
subprocess.run(['git','clone','--quiet','https://github.com/changxinjiresearch/PCC.git',str(repo)],check=True)
subprocess.run(['git','-C',str(repo),'checkout','--quiet','6588299'],check=True)
frozen=next(p.parent for p in Path('/kaggle/input').rglob('FULL_RUN_STATUS.json') if json.loads(p.read_text()).get('completed_p0_cases')==40)
out=Path('/kaggle/working/PCC_INTERNAL_COMPLETION_2026_SMOKE')
base=[sys.executable,'-m','experiments.run_internal_completion_2026','--frozen-root',str(frozen),'--output-root',str(out),'--config',str(repo/'configs/pcc_internal_completion_2026.yaml'),'--case-limit','1']
for family in ['mechanism','imperfect','targets']:
    subprocess.run(base+['--family',family],cwd=repo,check=True)
(out/'SMOKE_COMPLETE.json').write_text(json.dumps({'status':'COMPLETE','families':['mechanism','imperfect','targets'],'cases_each':1,'predictor_retrained':False,'p0_regenerated':False},indent=2)+'\n')
```
