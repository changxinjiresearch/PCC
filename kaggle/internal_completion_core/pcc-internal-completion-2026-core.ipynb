```python
from pathlib import Path
import subprocess,sys,json
repo=Path('/tmp/PCC')
subprocess.run(['git','clone','--quiet','https://github.com/changxinjiresearch/PCC.git',str(repo)],check=True)
subprocess.run(['git','-C',str(repo),'checkout','--quiet','941eec6'],check=True)
frozen=next(p.parent for p in Path('/kaggle/input').rglob('FULL_RUN_STATUS.json') if json.loads(p.read_text()).get('completed_p0_cases')==40)
out=Path('/kaggle/working/pcc_internal_completion_2026')
base=[sys.executable,'-m','experiments.run_internal_completion_2026','--frozen-root',str(frozen),'--output-root',str(out),'--config',str(repo/'configs/pcc_internal_completion_2026.yaml')]
for family in ['identity','mechanism','shuffled','targets']:
    subprocess.run(base+['--family',family],cwd=repo,check=True)
(out/'CORE_COMPLETE.json').write_text(json.dumps({'status':'COMPLETE','families':['identity','mechanism','shuffled','targets'],'cases':40,'predictor_retrained':False,'p0_regenerated':False},indent=2)+'\n')
```
