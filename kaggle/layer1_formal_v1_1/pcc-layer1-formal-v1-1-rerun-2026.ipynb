```python
from pathlib import Path
import json,subprocess,torch
assert torch.cuda.is_available(), 'GPU_REQUIRED'
assert torch.cuda.get_device_capability(0)==(6,0)
repo=Path('/tmp/PCC')
subprocess.run(['git','clone','--quiet','https://github.com/changxinjiresearch/PCC.git',str(repo)],check=True)
subprocess.run(['git','-C',str(repo),'checkout','--quiet','7d5c482'],check=True)
nb=json.loads((repo/'archive/pcc-experiments-original.ipynb').read_text())
def clean(i): return '\n'.join(x for x in ''.join(nb['cells'][i]['source']).splitlines() if not x.lstrip().startswith('!'))
exec(compile(clean(77),'archive-cell-77','exec'),globals())
exec(compile(clean(78),'archive-cell-78-formal-v1.1','exec'),globals())
d=Path('/kaggle/working/pcc_independent_baseline/stage2_pcc_guided_student_learning/layer1_current_segmentation_FORMAL_v1_1_15epoch_sensitivity')
m=__import__('pandas').read_csv(d/'layer1_FORMAL_v1_1_15epoch_case_metrics.csv')
assert len(m)==40 and m.case_id.nunique()==40
assert len(list((d/'checkpoints').glob('*.pt')))==10
(d/'LAYER1_V1_1_COMPLETION.json').write_text(json.dumps({'status':'COMPLETE','cases':40,'folds':5,'gpu_required':True,'source_cells':[77,78],'repository_commit':'7d5c482'},indent=2)+'\n')
```
