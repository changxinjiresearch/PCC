```python
from pathlib import Path
import json, subprocess, sys

commit = 'e13ee74'
repo = Path('/tmp/PCC')
subprocess.run(['git','clone','--quiet','https://github.com/changxinjiresearch/PCC.git',str(repo)],check=True)
subprocess.run(['git','-C',str(repo),'checkout','--quiet',commit],check=True)
status_files = list(Path('/kaggle/input').rglob('FULL_RUN_STATUS.json'))
assert len(status_files) == 1, status_files
frozen_root = status_files[0].parent
output = Path('/kaggle/working/pcc_downstream_v8')
command = [sys.executable, str(repo/'experiments/run_pcc_downstream.py'), '--family','sensitivity','--frozen-root',str(frozen_root),'--output-root',str(output)]
environment = dict(__import__('os').environ); environment['PYTHONPATH'] = str(repo)
subprocess.run(command,check=True,env=environment)
status=json.loads((output/'12_sensitivity/SENSITIVITY_STATUS.json').read_text())
assert status == {'status':'COMPLETE','cases':40,'configurations':8,'rows':320,'failed_cases':0,'predictor_retrained':False,'frozen_source_status_sha256':status['frozen_source_status_sha256']}
print(json.dumps(status,indent=2))
```
